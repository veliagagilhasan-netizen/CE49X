## Hints

### Gutenberg-Richter Fit

To compute the Gutenberg-Richter relationship:

```python
# Sort unique magnitudes
mags = np.sort(df['magnitude'].values)
# For each magnitude M, count earthquakes >= M
unique_mags = np.arange(mags.min(), mags.max(), 0.1)
cumulative_counts = np.array([np.sum(mags >= m) for m in unique_mags])
# Fit line to log10(counts) vs magnitude (only where count > 0)
mask = cumulative_counts > 0
coeffs = np.polyfit(unique_mags[mask], np.log10(cumulative_counts[mask]), 1)
b_value = -coeffs[0]  # b-value is the negative slope
```

### Colormaps

**Good choices** for earthquake risk data:
- `YlOrRd` — yellow to red, intuitive for risk (low → high)
- `Reds` or `OrRd` — single-hue sequential
- `inferno` or `magma` — perceptually uniform, works in grayscale
- `viridis` — perceptually uniform, good for depth data

**Avoid:** `jet`, `rainbow`, `hsv` — these create false boundaries and are not accessible to colorblind viewers.

fig = plt.figure(figsize=(16, 10))

### Deliverable 1: Data Collection & Documentation

````python
# === Dataset 1: Earthquake Catalog ===
# NOTE: For a real submission, replace this example with
# an actual Marmara-region catalog from AFAD/USGS and
# update the source metadata below accordingly.
# Source: Example synthetic catalog inspired by AFAD/USGS formats
# Date accessed: 14 March 2026
# Method: Programmatically generated for demonstration

np.random.seed(49)

# Create a synthetic earthquake catalog around the Marmara region
n_events = 350
years = np.random.randint(1900, 2026, size=n_events)
latitudes = np.random.uniform(40.2, 41.3, size=n_events)
longitudes = np.random.uniform(27.0, 30.5, size=n_events)
depths_km = np.random.uniform(5, 30, size=n_events)
magnitudes = np.round(np.random.normal(loc=4.2, scale=0.7, size=n_events), 1)

# Clip magnitudes to a reasonable range
magnitudes = np.clip(magnitudes, 2.0, 7.5)

# Add a few named historical events explicitly so we can annotate them
named_events = pd.DataFrame({
    'year': [1894, 1999, 1999],
    'latitude': [41.0, 40.70, 40.74],
    'longitude': [28.9, 29.91, 31.21],
    'depth_km': [15.0, 17.0, 15.0],
    'magnitude': [7.0, 7.6, 7.2],
    'event_name': ['1894 Istanbul', '1999 Izmit', '1999 Duzce']
})

random_events = pd.DataFrame({
    'year': years,
    'latitude': latitudes,
    'longitude': longitudes,
    'depth_km': depths_km,
    'magnitude': magnitudes,
    'event_name': [''] * n_events
})

df_quakes = pd.concat([random_events, named_events], ignore_index=True)

print('Earthquake catalog records:', len(df_quakes))
df_quakes.head(), df_quakes.describe(include='all')
````

````python
# === Dataset 2: Building / Population Data ===
# NOTE: For a real submission, you should replace this
# illustrative dataset with official values from
# IBB/TUIK or another documented source.
# Source: Example synthetic district-level dataset
# Date accessed: 14 March 2026
# Method: Manually constructed for demonstration

istanbul_districts = [
    'Avcilar', 'Bakirkoy', 'Besiktas', 'Kadikoy', 'Fatih',
    'Kartal', 'Zeytinburnu', 'Esenler', 'Maltepe', 'Uskudar',
    'Beyoglu', 'Kucukcekmece'
]

population = [
    430000, 230000, 180000, 480000, 430000,
    470000, 290000, 450000, 520000, 530000,
    240000, 760000
]

area_km2 = [
    42, 30, 11, 25, 15,
    38, 12, 18, 53, 36,
    9, 37
]

# Approximate building counts split by pre-/post-1999 stock
buildings_pre1999 = [
    38000, 18000, 12000, 20000, 26000,
    22000, 17000, 31000, 21000, 24000,
    14000, 42000
]

buildings_post1999 = [
    22000, 14000, 10000, 26000, 15000,
    23000, 9000, 12000, 24000, 26000,
    11000, 28000
]

df_districts = pd.DataFrame({
    'district': istanbul_districts,
    'population': population,
    'area_km2': area_km2,
    'buildings_pre1999': buildings_pre1999,
    'buildings_post1999': buildings_post1999
})

# Derived metrics
df_districts['pop_density'] = df_districts['population'] / df_districts['area_km2']
df_districts['prop_pre1999'] = df_districts['buildings_pre1999'] / (
    df_districts['buildings_pre1999'] + df_districts['buildings_post1999']
)

print('Building/population records:', len(df_districts))
df_districts.head(), df_districts.describe(include='all')
````

````python
# === Dataset 3: Seismic Hazard / Vulnerability Data ===
# NOTE: In practice, this should be based on AFAD/IBB
# hazard maps or a published risk study. Here we
# construct an illustrative dataset aligned with the
# districts defined in df_districts.
# Source: Example synthetic PGA / risk values
# Date accessed: 14 March 2026
# Method: Manually constructed for demonstration

# Approximate district centre coordinates (very rough)
lat_center = [
    40.98, 40.97, 41.04, 40.99, 41.02,
    40.89, 40.99, 41.03, 40.93, 41.03,
    41.04, 41.00
]

lon_center = [
    28.72, 28.86, 29.01, 29.03, 28.95,
    29.19, 28.90, 28.88, 29.11, 29.03,
    28.97, 28.80
]

# Hypothetical peak ground acceleration values (g)
pga_g = [
    0.40, 0.32, 0.28, 0.30, 0.36,
    0.34, 0.38, 0.35, 0.31, 0.29,
    0.33, 0.37
]

# Simple combined risk index: population density × PGA
hazard_base = pd.DataFrame({
    'district': istanbul_districts,
    'lat': lat_center,
    'lon': lon_center,
    'pga_g': pga_g
})

# Join with district data so we can derive a risk metric
df_hazard = hazard_base.merge(df_districts[['district', 'pop_density']], on='district')
df_hazard['risk_index'] = df_hazard['pga_g'] * df_hazard['pop_density']

print('Hazard/vulnerability records:', len(df_hazard))
df_hazard.head(), df_hazard.describe(include='all')
````

### Deliverable 3: Magnitude-Frequency Analysis

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Subplot 1: Magnitude histogram with KDE ---
ax1 = axes[0]

# Using 0.2-magnitude bins because the synthetic catalog
# reports magnitudes to 0.1 precision and we want to avoid
# an overly noisy histogram while still resolving structure.
min_mag = df_quakes['magnitude'].min()
max_mag = df_quakes['magnitude'].max()
bins = np.arange(np.floor(min_mag), np.ceil(max_mag) + 0.2, 0.2)

sns.histplot(
    df_quakes['magnitude'],
    bins=bins,
    kde=True,
    stat='count',
    color='steelblue',
    edgecolor='white',
    ax=ax1
)

mean_mag = df_quakes['magnitude'].mean()
median_mag = df_quakes['magnitude'].median()

ax1.axvline(mean_mag, color='red', linestyle='--', label=f"Mean = {mean_mag:.2f}")
ax1.axvline(median_mag, color='green', linestyle=':', label=f"Median = {median_mag:.2f}")

ax1.set_xlabel('Magnitude (Mw)')
ax1.set_ylabel('Number of events')
ax1.set_title('Magnitude Distribution of Marmara Earthquakes')
ax1.legend()

# --- Subplot 2: Gutenberg-Richter plot ---
ax2 = axes[1]

mags = np.sort(df_quakes['magnitude'].values)
unique_mags = np.arange(mags.min(), mags.max(), 0.1)
cumulative_counts = np.array([np.sum(mags >= m) for m in unique_mags])

mask = cumulative_counts > 0
log_counts = np.log10(cumulative_counts[mask])
mag_for_fit = unique_mags[mask]

# Fit linear regression: log10(N) = a - b M
coeffs = np.polyfit(mag_for_fit, log_counts, 1)
a_val, b_slope = coeffs
b_value = -b_slope

fit_line = np.polyval(coeffs, mag_for_fit)

ax2.plot(unique_mags[mask], log_counts, 'o', ms=4, label='Catalog (cumulative)', color='black')
ax2.plot(mag_for_fit, fit_line, '-', color='crimson', label=f'Fit (b = {b_value:.2f})')

ax2.set_xlabel('Magnitude threshold M')
ax2.set_ylabel(r'$\log_{10} N(\geq M)$')
ax2.set_title('Gutenberg-Richter Relation for Synthetic Catalog')
ax2.grid(True, linestyle='--', alpha=0.4)
ax2.legend()

plt.tight_layout()
plt.show()
````

#### Reflection

**1. Chart type justification (D2–D5):**

For the historical seismicity timeline (D2) I used a scatter plot with time on the x-axis and magnitude on the y-axis because, according to the encoding hierarchy from the theory notebook, position is the most accurate channel for quantitative variables; this makes temporal trends in magnitude immediately readable. Marker size encodes magnitude redundantly to draw attention to larger events, and color encodes depth using a perceptually uniform sequential colormap (`viridis`), following the recommendation to avoid rainbow colormaps. For the magnitude histogram and Gutenberg–Richter plot (D3), a histogram plus KDE is the standard way to summarize a 1D distribution while controlling bin width, and the log–linear Gutenberg–Richter plot is a classic example of using transformation (log scale) to reveal a power-law relationship. For the district vulnerability plots (D4), bar charts were chosen for ordered categorical comparisons (districts on the x-axis, density or proportion on the y-axis), which aligns with the chart selection framework for comparing magnitudes across categories; districts are sorted by the metric of interest to avoid the “alphabetical ordering” pitfall. For the hazard visualization (D5), a spatial scatter map was used because spatial data should be shown in spatial coordinates; color encodes PGA and size encodes a derived risk measure, consistent with the principle that more important variables should be mapped to more salient channels.

**2. Audience adaptation (seismologists vs. municipal council):**

For a seismologist audience, I would increase the technical detail and show more of the underlying uncertainty. For example, in the Gutenberg–Richter plot I would include confidence intervals on the fitted line, the magnitude of completeness, and residual plots, and I would use precise seismic notation instead of a narrative title. In the hazard visualizations, I would replace qualitative labels like “higher risk” with exact PGA ranges, site classes, and perhaps spectral accelerations at different periods, because seismologists are comfortable with these parameters. For the municipal council, I simplified the visuals, used intuitive color scales (e.g., yellow-to-red for risk), and wrote titles as short messages that answer a policy-relevant question (e.g., which districts should be prioritized) instead of focusing on methodological detail.

**3. Self-critique (weakest visualization + improvement plan):**

The weakest visualization in this notebook is likely the simple scatter-based hazard map, which approximates district-level hazard with single points and does not fully capture intra-district variability. This relates to the “oversimplification / missing context” mistake from the theory notebook: the viewer might over-interpret point locations as precise boundaries. To improve it, I would replace the point representation with a proper choropleth using actual district polygons, refine the color scale using perceptually uniform steps anchored to engineering design thresholds, and add clearer annotations about limitations (e.g., that values are aggregated district means). I would also improve geographic context by adding coastlines, major fault traces, and key infrastructure so that policy-makers can better connect the hazard pattern to familiar locations.
